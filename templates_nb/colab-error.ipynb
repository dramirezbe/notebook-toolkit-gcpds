```python
!pip install -q uv

import os
import sys
import subprocess
from IPython.core.magic import register_cell_magic

IS_COLAB = "google.colab" in sys.modules
IS_KAGGLE = "KAGGLE_KERNEL_RUN_TYPE" in os.environ or "KAGGLE_URL_BASE" in os.environ
IS_DATALORE = any(k.startswith("DATALORE") for k in os.environ) or os.path.exists("/data/notebook_files")
IS_DEEPNOTE = any(k.startswith("DEEPNOTE") for k in os.environ) or os.path.exists("/work")
IS_CLOUD_NOTEBOOK = IS_COLAB or IS_KAGGLE or IS_DATALORE or IS_DEEPNOTE

TARGET_PYTHON = "3.11.10"
VENV_PATH = "/tmp/py311_env"
REPO_URL = "https://github.com/dramirezbe/notebook-toolkit-gcpds.git"
TMP_REPO_DIR = "/tmp/notebook-toolkit-gcpds"

# 1. Fetch Python 3.11.10 and build environment
!uv python install {TARGET_PYTHON}
!uv venv --python {TARGET_PYTHON} --clear {VENV_PATH}

# 2. Clone repository and install dependencies
!rm -rf {TMP_REPO_DIR} && git clone --depth 1 {REPO_URL} {TMP_REPO_DIR}
!cd {TMP_REPO_DIR} && uv pip install --python {VENV_PATH}/bin/python . numpy pandas scipy scikit-learn

# 3. Register fixed magic using subprocess
@register_cell_magic
def py311(line, cell):
    script_path = "/tmp/temp_cell.py"
    with open(script_path, "w") as f:
        f.write(cell)

    python_bin = f"{VENV_PATH}/bin/python"
    if not os.path.exists(python_bin):
        print(f"Error: Python binary not found at {python_bin}")
        return

    result = subprocess.run(
        [python_bin, script_path],
        capture_output=True,
        text=True
    )

    if result.stdout:
        print(result.stdout, end="")
    if result.stderr:
        print(result.stderr, end="", file=sys.stderr)

print("\nSetup complete! You can now use %%py311.")
```
